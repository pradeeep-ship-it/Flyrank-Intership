## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

```python
## 2. Target or Proxy

**Target Column:** `needs_refresh`

The goal is to predict whether a webpage should be refreshed.

Since the dataset does not contain a direct label, we define a proxy target based on webpage performance.

Target values:

- **1** → Refresh Needed
- **0** → No Refresh Needed

A webpage may need refreshing if it shows declining traffic, lower engagement, or reduced clicks over time.

This prediction helps the content team prioritize which webpages should be updated.
```

## 3. Success metric

*One metric you can defend. What number means 'good'?*

```python
## 3. Success Metric

**Primary Metric:** F1 Score

F1 Score is chosen because it balances Precision and Recall.

The objective is to correctly identify webpages that require refreshing while minimizing unnecessary refresh recommendations.

A higher F1 Score indicates better model performance.
```

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

```python
## 5. Why ML Beats a Fixed Rule Here

A simple rule such as **"Refresh every page whose traffic drops by 20%"** is not sufficient because webpage performance depends on many factors.

These include:

- Traffic trend
- User engagement
- Click-through rate (CTR)
- Content age
- Seasonal changes
- Topic popularity

Machine Learning can learn complex relationships among these features and make better predictions than a single threshold-based rule.

This helps the content team prioritize which webpages should be refreshed first.
```
